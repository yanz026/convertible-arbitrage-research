## Understanding Convertible Bonds


* **Definition**: A convertible bond (“convert”) is a corporate bond with an embedded call option on the issuer’s stock, giving the holder the right, but not the obligation, to convert it into a predetermined number of common shares at a fixed conversion price and at specific times. This hybrid security offers bond-like interest payments while also providing equity upside potential. It behaves like a traditional corporate bond when the stock price is low (out-of-the-money option) and like equity when the stock price is high (in-the-money option). Key features of the bond include:


 * **Conversion Price**: The fixed stock price at which the bond converts into shares. This functions as the strike price of the embedded equity option, meaning the bondholder will typically convert if the stock price is at or above the conversion price (i.e., the option is in-the-money).


 * **Conversion Ratio**: The number of shares of common stock the bond can be converted into. It is determined by dividing the face value of the bond by the conversion price, where the face value is typically $1,000 per bond.


 * **Coupon, Maturity, Face Value**: Like a traditional corporate bond, a convert has a fixed coupon rate, maturity date, and face value. However, its value is influenced by both its debt characteristics and the conversion option.






**Why do firms issue converts?**
 1. **Low coupons:** Companies (especially growth firms) issue converts to raise capital cheaply. The embedded option enables issuers to offer lower coupons than traditional bonds. The typical U.S. convert issuance ranges from $200 million to $500 million.


 2. **Strategic Benefits**: Converts allow firms to capitalize on equity volatility, delay dilution by issuing equity at a premium, and gain tax advantages. Their flexible structure enables issuers to tailor their bond issuance to their specific business needs.




**Why do people buy converts?**
 1. **Enhanced Risk-Return Profile**: Converts offer fixed income with equity upside, downside protection, and low correlation with other assets, making them a valuable tool for portfolio diversification. Additionally, converts often provide a higher yield than simply holding the underlying stock.


 2. **Arbitrage opportunities**: Arbitrageurs profit from discrepancies between a convertible bond’s theoretical value and market price, driven by volatility, credit risk, interest rates, dividends, and supply-demand imbalances.




**Example Walkthrough: Investing in a Convertible Bond**


Suppose you have a strong conviction that a growth company will perform well over the next few years.


The company's stock is currently $5, and you expect it to rise as the company improves.
   - They issue a convertible bond with:
       - $1,000 face value 
       - $200 conversion price 
       - 1% coupon paid semiannually 
       - 5-year maturity 


Key Features: 
Like a traditional bond, it pays a $5 coupon every 6 months (1% annually on a $1,000 face value). 
The conversion ratio is 5 ($1,000 face value ÷ $200 conversion price = 5 shares per bond). 


What Happens at Maturity (Year 5)? The bondholder chooses whether to convert the bond into equity or redeem it for cash:

If the stock price is below $200 → The bondholder keeps the bond and redeems it for $1,000 face value, ensuring downside protection. 


Conclusion: The bondholder benefits from equity upside while having limited downside risk, guaranteed by the bond floor (face value of $1,000).

## Strategy Rationale: Convertible Bond Arbitrage
 
Our general trade idea is to systematically capture risk premia in convertible bonds by taking long positions in convertible bonds while simultaneously shorting the underlying stock to hedge equity exposure. The strategy seeks to profit from structural inefficiencies, such as liquidity constraints, volatility risk premia, and delayed price adjustments in an OTC market. By identifying instances where a convertible bond's implied volatility is misaligned with the realized volatility of the underlying equity, as well as leveraging inefficiencies caused by fragmented price discovery and execution frictions, we aim to generate risk-adjusted returns in a market that remains largely dominated by manual trading.


### Why do we think there is a trading opportunity? (i.e. Why do we think convertible bonds are mispriced?)
 1. **OTC Product with Delayed Price Discovery:** Convertible bonds primarily trade over-the-counter (OTC), meaning they lack the transparency and real-time price discovery of exchange-listed securities. Unlike common stocks, which update prices in real time, converts experience delayed price discovery due to fragmented trading activity and reliance on dealer quotes rather than continuous market order flow. This price opacity creates opportunities for mispricing, particularly when fundamental changes in credit risk, interest rates, or stock volatility are not immediately reflected in bond prices.


 2. **Complex Valuation & Market Segmentation:** Convertible bonds are hybrid instruments combining a traditional fixed-income bond with an embedded equity option. This dual nature makes valuation challenging, as it requires incorporating both a bond component influenced by credit risk and interest rate movements and an equity option component dependent on stock price fluctuations and volatility. While the Black-Scholes model laid the foundation for modern option pricing, Brennan and Schwartz (1977) later expanded these models to include convertible bonds. Traditional bond investors often undervalue the option component, while equity traders tend to overlook credit risk, leading to systematic pricing inefficiencies. Academic research, including Ammann et al. (2003), supports this, showing that CBs frequently trade below theoretical value due to investor segmentation and pricing complexity.


 3. **Liquidity Constraints Create Inefficiencies:** The convertible bond market is small, with fewer participants compared to traditional fixed-income and equity markets. Unlike common stocks which attract a diverse range of retail and institutional investors, the majority of convertible bond traders are hedge funds, institutional credit desks, and dedicated arbitrage funds. This limited participation contributes to structural illiquidity. Large orders can significantly impact prices, and forced liquidations—such as during the 2008 financial crisis—cause CBs to trade well below fair value. Agarwal et al. (2011), highlight how liquidity crises exacerbate pricing inefficiencies. For example, during the 2008 financial crisis, many convertible arbitrage funds faced margin calls, forcing them to liquidate bonds at well below fair value. These pricing dislocations persisted until market stability returned, offering lucrative opportunities for well-capitalized investors. Investors who recognized this opportunity and acquired distressed convertibles benefited from strong price recoveries as liquidity returned.


  4. **Volatility Mispricing & Arbitrage:** Convertible bond prices are highly sensitive to implied volatility assumptions, which traders use to estimate the fair value of the embedded option. However, realized volatility often deviates from implied levels, leading to valuation mismatches that arbitrageurs can exploit. Hedge funds specializing in volatility arbitrage capitalize on these inefficiencies by buying convertibles when implied volatility is artificially low, expecting realized volatility to increase and shorting convertibles when implied volatility is overstated, anticipating a decline in realized volatility. Moreover, corporate events such as M&A activity, earnings surprises, or regulatory changes can cause sudden volatility spikes, making previous pricing models obsolete and creating short-term arbitrage opportunities. If a stock’s realized volatility increases, but the convertible bond price does not adjust accordingly, it may indicate that the convertible is undervalued relative to its new risk profile.


  5. **Arbitrage Frictions & Market Distortions:** While convertible arbitrage strategies seek to exploit pricing inefficiencies, several frictions can distort fair value calculations: 1. Stock Borrowing Constraints: Shorting the underlying stock to hedge a convertible bond position may be expensive or unavailable, limiting arbitrage effectiveness. 2. Short Squeezes: If multiple arbitrage funds short the same stock, unexpected price spikes can force them to cover at elevated prices, inflating convertible bond values temporarily. 3. Corporate Actions: Events like dividend hikes, stock buybacks, or dilution can impact convertible pricing unexpectedly, creating arbitrage opportunities. Loncarski et al. (2009) found that short interest in stocks increases significantly around convertible bond issuance, suggesting that arbitrageurs actively hedge their positions. However, unexpected shifts in corporate policy can disrupt these trades. If a company suddenly increases dividends, its stock price might drop. However, convertible bond models may not immediately adjust, creating a temporary mispricing that sophisticated traders can exploit.

## Our Bond Pricing Model:
* **Explaining the Bond + Option Model:** Convertible bonds are hybrid securities that combine a traditional bond with an embedded call option on the issuer’s stock. The bond + option pricing model values a convertible bond by separately assessing the fixed-income component and the equity option component, then summing their values to determine the fair price of the convertible.


   * Pricing the bond component:
       * Compute the present value of future coupon payments and the principal repayment at maturity.
       * Discount using a credit-adjusted discount rate, which accounts for the issuer’s default risk.
       * If the issuer defaults, the bondholder receives a recovery amount (typically based on seniority in the capital structure).
       * Formula: $$\sum_{t=1}^{T} \frac{C}{(1+r)^t} + \frac{F}{(1+r)^T}$$ where
           * C = Coupon payment,
           * r = credit-adjusted discount rate (which changes at every period)
           * T = number of years to maturity
           * F = face value of the bond 


   * Pricing the option component:
       * The conversion feature gives the bondholder the right to convert the bond into a fixed number of shares at a predetermined conversion price.
       * This embedded call option is valued using option pricing models, such as Black-Scholes or more advanced approaches like binomial trees or Monte Carlo simulations to account for early conversion possibilities.
       * The option’s value is influenced by factors such as:
           * Stock price (higher prices increase option value)
           * Volatility (higher volatility increases the likelihood of conversion)
           * Time to maturity (longer time frames allow more price movement)
           * Dividends (higher dividends decrease option value, as stockholders receive dividends but bondholders do not)
       * Formula: $$C = S_0 e^{-qT} N(d_1) - K e^{-rT} N(d_2)$$ where $$d_1 = \frac{\ln(S_0/K) + (r - q + \sigma^2/2)T}{\sigma \sqrt{T}}$$ and $$d_2 = d_1 - \sigma \sqrt{T}$$ where
           * S = spot price
           * K = strike price
           * sigma = volatility thesis
           * r = risk-free rate
           * T = time to maturity (in years)


   * Accounting for Early Conversion and Call/Out Features: Convertible bonds often include adjustments for early conversion and call/put features such as:
       * Issuer Call Provisions – Allowing the issuer to redeem the bond early at a set price.
       * Investor Put Provisions – Allowing bondholders to sell the bond back to the issuer at a fixed price before maturity.
       * Forced Conversion – If the stock price reaches a certain level, the issuer may force conversion, effectively capping the bond’s upside.




* **Our Usage of the Model:** The two core components that differentiate a bond + option pricing model’s theoretical price from the market price are **1. the credit spread assumption** and **2. the volatility assumption**. These inputs are set by the user based on their thesis regarding the company’s ability to meet its debt obligations and their outlook on the company’s future stock volatility. 


   * To form a credit assumption, the user must develop a thorough understanding of the company’s ability to meet its liabilities by analyzing its fundamentals. This includes examining its debt stack and financial statements (balance sheet, cash flow statement, income statement), attending shareholder meetings, reading news articles, and conducting other forms of due diligence. Only after understanding company fundamentals can a user make an informed decision about the appropriate credit spread. A higher credit spread lowers the model’s theoretical value of the convertible bond, while a lower credit spread increases its theoretical value. 


       Given the scope and timeframe of our project, we did not have the bandwidth to implement an automated bot for parsing financial statements or conducting sentiment analysis on news related to the tickers in our universe. To simplify the process, we use CDS spreads as our credit spread assumption, as they represent the market consensus on the company’s creditworthiness. To compute the credit-adjusted discount rate, we add the company’s CDS spread to the risk-free rate (the zero-coupon Treasury rate of the same maturity as the convertible bond). While this simplifies pricing, we acknowledge that since we are using a market-driven credit metric, deviations in our model’s theoretical bond value will primarily stem from our volatility thesis rather than the credit assumption. 


   * To compute our volatility thesis, we use a Weighted Root Mean Square (WRMS) of both market-implied volatility and historical realized volatility:
       - Market-implied volatility is extracted from option prices and represents the market’s expectation of future volatility. 
       - Historical realized volatility is derived from past stock price movements and reflects actual fluctuations. 
       - Since both measures are imperfect but informative, combining them provides a more balanced and reliable volatility estimate. 


       We use WRMS instead of a simple average because it emphasizes large deviations in volatility. Volatility is not linear—extreme price swings have an outsized impact on convertible bond valuations due to the convexity of the embedded option. WRMS naturally gives greater weight to higher volatility values, which is crucial when pricing convertibles because: 
       - Higher stock volatility increases the option value embedded in the bond. 
       - The impact of volatility is nonlinear, as seen in pricing models like Black-Scholes. 


       Additionally, using WRMS prevents underestimation of volatility that could occur with simple averaging. A simple arithmetic mean could underestimate volatility when there are large spikes in implied or realized volatility. Since WRMS squares volatility values before averaging, it preserves the impact of extreme movements, making it a better estimator for pricing the convertible’s embedded option. Ultimately, WRMS ensures that the volatility input into our pricing models is robust, reducing the risk of pricing errors.
* **Justification of Why We Use the Bond + Option Model:** The Bond + Option Model is a widely used framework for pricing convertible bonds due to its robustness and computational efficiency. It provides a structured breakdown of a convertible’s value by separately assessing its bond and option components, ensuring a transparent decomposition and fair valuation. Its structured transparency, robustness, and computational efficiency make it a practical tool for traders, investors, and arbitrageurs. We use this model instead of more complex alternatives because:


   * **Accuracy vs. Simplicity Tradeoff**: While more sophisticated pricing models exist (e.g., binomial trees, finite difference methods, or Monte Carlo simulations), they require significantly more computational power and are more difficult to interpret. The Bond + Option Model is widely accepted in practice because it balances accuracy with simplicity, often yielding valuations that are within 99% accuracy of 
more complex models while being much faster to compute.


   * **Computational Efficiency for Real-Time Trading**: Convertible arbitrage and market-making strategies require quick, real-time valuations. The Bond + Option Model is efficient enough to be used in live pricing, while models like binomial trees or Monte Carlo simulations require significantly more time to generate estimates.
   * Flexibility in Incorporating Credit and Volatility Assumptions: This model allows us to adjust for credit risk using a company’s CDS spreads instead of relying on complex default probability models. It also allows us to customize volatility assumptions, which is critical given that a convertible bond’s value is highly sensitive to stock price volatility.


   * **Practical Relevance in Trading and Risk Management**: Most market practitioners (traders, hedge funds, and investment firms) rely on this model for relative value analysis. Since convertibles often trade based on relative mispricing, it is more important to have a consistent, fast, and comparable model rather than a theoretically perfect one.


The Bond + Option Model remains the preferred choice for pricing convertible bonds due to its accuracy, computational efficiency, and practical flexibility. While more complex models may provide slight improvements in theoretical accuracy, they are often impractical for real-world trading applications. Our usage of this model ensures we can efficiently evaluate relative value opportunities while maintaining a robust pricing framework that aligns with how convertibles are traded in the market.


* **Acknowledging Potential Flaws of the Bond + Option Model**:
   * Most convertible bonds include an American-style option, allowing conversion at any time before maturity. However, our Bond + Option model assumes a European-style option and applies the Black-Scholes model for pricing. This assumption is valid when the underlying stock does not pay dividends, as early exercise of an American call option is never optimal in such cases. Consequently, an American call option behaves identically to a European call option, making the simplification reasonable. Since most underlying stocks of convertibles do not pay dividends, this approach generally holds.
       * However, if the underlying stock does pay dividends, treating the convertible bond as a simple bond + European call option package becomes inaccurate. In such cases, early exercise before ex-dividend dates can be optimal, making the American-style feature relevant. For dividend-paying stocks, a more sophisticated pricing model—such as binomial trees or finite difference methods—is needed to properly capture early exercise behavior.
       * To try to account for the effect of dividends, we will adjust the Black-Scholes Model to include the dividend yield (i.e. the Black-Scholes-Merton model). However, we acknowledge that even accounting for dividend yield q does not quite capture the dynamics of the American option.
  
   * Similarly, convertible bond contracts can include additional terms such as issuer calls, investor puts, and forced conversions, all of which impact pricing. However, valuing a convertible bond solely as a vanilla corporate bond or using Black-Scholes for the option component does not account for the complex pricing dynamics introduced by these clauses. To accurately model convertible bond prices, more advanced methods such as binomial trees, finite difference techniques, or Monte Carlo simulations would be required. While more accurate, as explained above, this method would be more computationally and time-intensive.

## Strategy Details


   * **Identifying & Entering Underpriced Convertible Bonds:** We buy a convertible bond when we believe it is undervalued, meaning the market is underestimating volatility. This occurs when implied volatility is lower than our expected volatility or when the market price is below our model’s theoretical value. We take a long position when a convertible bond is trading between 0% and 4% cheap relative to our model’s valuation. 


   * **Why Do We Cap Cheapness at 4%?** We set a 4% boundary because while our model is robust, it is not perfect and is subject to potential pricing errors. Research by Ammann et al. (2003) suggests that convertible bonds typically trade at a 2-5% discount, supporting our decision to cap reasonable cheapness at 4%. If the model indicates a discount greater than 4%, we assume it is overestimating the bond’s cheapness and do not act. However, if the model suggests a 0-4% discount, we consider it within reason and take a long position accordingly. This approach helps us capture pricing inefficiencies while mitigating the risk of model miscalculations.


   * Managing Risk Through Hedging
      **Delta Hedging to Stay Market Neutral:** When we take a long position in a convertible bond, we simultaneously short the underlying stock to hedge our delta risk. Delta measures how much the convertible bond's price moves in relation to changes in the stock price due to its embedded option. By shorting the stock in proportion to the bond’s delta, we ensure that our position remains directionally neutral to equity price movements.


      **Duration Hedging to Offset Interest Rate Risk:** Since convertible bonds also have a fixed-income component, they are subject to interest rate risk, where rising rates can lead to lower bond prices. To offset this risk, we hedge our duration exposure with Treasuries to minimize sensitivity to interest rate fluctuations.


   * **Adjusting Hedges Dynamically Throughout the Trade:** Throughout the life of the trade, we continuously adjust our delta and duration hedges to maintain market neutrality and reduce exposure to directional stock and interest rate movements. In maintaining our delta hedge, we are able to earn some additional PnL from our unhedged risks, such as gamma and vega. For example, since our gamma is positive, when the stock price increases, we need to short more stock, and when the stock price decreases, we need to buy more stock. Thus, when we are readjusting our stock position to remain delta hedged, we are selling the stock high and buying low.


   * **Profit Taking: Exiting When Market Price Converges:** When the convertible bond’s market price converges to its theoretical value, we exit the position and realize profits. Specifically, if we believe vol is cheap, we go long the convertible bond and profit as our volatility thesis realizes (i.e. implied vol converges to realized vol, or equivalently, the market price converges to our model's theoretical value).


   * **Other Exit Scenarios: Maturity & Distressed Bonds:**: We will also close a position when the convertible bond matures, collecting $1,000 in principal per bond. Even if a bond diverges from our model’s theoretical value, we will continue to hold it, as it will ultimately converge to $1,000 at maturity, assuming the company does not default. This approach is on average the loss-minimizing and profit-maximizing decision, as it ensures we realize the bond’s full face value. In the event of a default, we will not sell the bond immediately. Since convertible bonds are senior unsecured debt, holding through the bankruptcy process is often preferable to selling at a distressed price. If a company files for bankruptcy, the bond’s price may drop significantly (e.g., from $100 to $20), but selling at such a steep discount often results in greater losses than waiting for the restructuring process. By holding, we may recover more value through the continuation process, as senior unsecured bondholders typically receive some recovery in restructuring, making it a better alternative than liquidating at distressed prices. Additionally, At the end of our data sample, we will forcibly close all positions and realize our PnL on them.

## Managing Risk 
A well-structured risk management framework is essential to executing a successful convertible bond arbitrage strategy. While many of the inherent risks can be actively hedged, others require careful monitoring and dynamic adjustments. Below, we outline the primary risks our strategy faces and how we mitigate them, as well as those that remain outstanding and must be accounted for in our overall portfolio management.  


### Risk Management


Effective risk management involves neutralizing key exposures while allowing us to take advantage of certain market inefficiencies. The two primary risks we hedge are duration risk and delta risk, both of which arise due to the dual nature of convertible bonds—part bond, part equity derivative.  


**Duration Risk**
Since convertible bonds contain a fixed-income component, they are sensitive to interest rate fluctuations. Rising interest rates can decrease bond prices, while falling rates can increase them.  


To insulate ourselves from this risk, we short Treasuries to offset changes in bond valuations due to shifts in the interest rate environment. However, because bond duration changes as time passes and rates fluctuate, we must continuously adjust our Treasury hedge to maintain a duration-neutral position.  


- If interest rates rise, bond prices drop, and duration may decrease, especially for callable convertibles.  
- If interest rates fall, bond prices rise, and duration may increase.  
- As time passes, a bond’s duration naturally declines due to the pull-to-par effect.  
- This means we must adjust our Treasury hedge over time to maintain a constant duration exposure.  


**Delta Risk**
A convertible bond’s embedded option introduces exposure to movements in the underlying stock price.  


To hedge against directional stock price moves, we short the underlying stock when long the convertible bond. Conversely, if shorting a convertible bond, we go long the stock. Because delta is dynamic and changes as the stock price moves, our hedge must be frequently rebalanced to maintain a delta-neutral position.  


- Delta represents how much the option price moves relative to the stock price.  
- As the stock price rises, delta typically increases for a long call option.  
- As the stock price falls, delta decreases, reducing equity exposure.  


Since delta is constantly changing, we can capitalize on market movements by dynamically adjusting our hedge.  


- If gamma is positive:  
  - When stock price increases, delta increases, requiring us to short more stock.  
  - When stock price decreases, delta decreases, requiring us to buy stock back.  
  - This results in buying low and selling high, generating additional returns over time.  


Additionally, by keeping certain Greek exposures unhedged—such as gamma and vega—we can potentially generate P&L from market fluctuations.  


**Liquidity Risk**
To mitigate liquidity risks inherent in the convertible bond market, we have implemented a two-pronged approach to ensure efficient trade execution and minimize market impact.  


First, we intentionally focus on trading the most liquid convertible bonds, selecting securities with higher trading volumes and tighter bid-ask spreads. By prioritizing convertibles that have a more active secondary market, we reduce the likelihood of execution delays and price slippage. This approach helps ensure that we can enter and exit positions more efficiently compared to less liquid convertibles.  


Second, we carefully size our positions to avoid moving the market, ensuring that our trades do not materially impact pricing or introduce unnecessary execution risks. By keeping position sizes within reasonable limits relative to daily traded volume, we minimize the risk of excessive slippage and maintain the ability to adjust or unwind trades without significant price distortions.  


Through this liquidity-aware trade selection and sizing strategy, we enhance execution efficiency, reduce transaction costs, and improve the robustness of our arbitrage strategy in real-world trading conditions.  




### Outstanding Risks  


Despite careful hedging, some risks cannot be fully eliminated and must be actively managed throughout the life of the trade.  


**Gamma/Vega Exposure**
Convertible bonds contain an embedded option whose value is directly influenced by market volatility. This introduces gamma and vega risk, which can impact strategy profitability under different volatility regimes. To capitalize on volatility-driven opportunities, we intentionally leave gamma and vega exposures unhedged, allowing us to benefit from favorable market movements. However, this also introduces risks that require active monitoring and adjustments.  


If implied volatility declines, the option component of the convertible bond loses value, reducing arbitrage opportunities. A sharp decline in realized volatility can also create a mismatch between implied and actual volatility, impacting hedge effectiveness.  


Conversely, if market volatility spikes, delta and gamma can shift rapidly, requiring more frequent rebalancing of the stock hedge. Sudden volatility surges can also make gamma hedging more costly, particularly if stock borrow rates rise or execution liquidity deteriorates.  


**Scenarios that Impact Convertible Bond Pricing**  
- If implied volatility declines, the embedded option loses value, reducing arbitrage opportunities.  
- If market stress causes volatility spikes, gamma hedging becomes more difficult and costly.  


**Stock Borrowing Cost Risk**
Since our strategy involves shorting the underlying stock, we rely on being able to borrow shares. If borrow availability becomes scarce, costs can spike, making the hedge prohibitively expensive.  


In extreme cases, borrow rates may become uneconomical, or shares may become unavailable altogether, forcing us to unwind trades prematurely. This risk is particularly relevant for small-cap convertibles or highly shorted stocks, where borrow costs can be volatile.  


**Credit Risk** 
While hedging the equity exposure of a convertible bond mitigates stock-related risks, it does not fully eliminate credit risk.  


If an issuer’s financial health deteriorates, credit spreads can widen, reducing the bond’s value. In the event of a default, bondholders are still exposed to losses, as the recovery value of the bond may be significantly lower than anticipated.  


Credit events impacting convertible arbitrage:  
- Falling credit quality increases the likelihood of default, reducing convertible bond value.  
- Forced conversions or restructurings can introduce unpredictable risks to arbitrage trades.

## Real World / Practical Considerations  


While this strategy has shown strong performance in our backtests and is well-supported by academic literature, there are several important factors to consider before making an investment that are difficult to model without paper trading or real trading.  


**Liquidity Concerns for the Short**
The literature indicates that when convertible bonds are first issued, many market participants attempt to short the corresponding equity simultaneously to hedge the newly issued convertible bond (Choi et al., 2009). If too many traders are trying to short the same security, it can lead to liquidity constraints and difficulties in sourcing shares for borrowing. This could result in higher borrow fees or even an inability to execute the hedge at desired levels.  


However, our strategy is less likely to encounter these issues, as we are not targeting equities immediately after a convertible bond is issued or announced. By focusing on arbitrage opportunities in seasoned convertible bonds, we reduce our exposure to these liquidity bottlenecks.  


**Liquidity Concerns for the Convertible Bond / OTC Trading**
As previously mentioned, convertible bonds trade over-the-counter (OTC) rather than on a centralized exchange. This creates several challenges:  
- Price Transparency: Unlike exchange-traded securities, convertible bonds do not have a consolidated order book, making price discovery difficult.  
- Limited Trading Volume: Some convertible bonds trade infrequently, leading to stale quotes and wider bid-ask spreads.  


To address these challenges, our strategy must account for potential delays in execution and price slippage when entering or exiting positions.  


**Market Impact – Consider Accumulation Algorithm**
Since convertible bonds are relatively illiquid, large trades can move the market, impacting execution prices. This is particularly relevant when establishing or unwinding large positions.


To minimize market impact, the strategy should incorporate:  
- Order slicing and execution algorithms (e.g., VWAP, TWAP) to spread out trades over time.  
- Dark pool or block trading mechanisms to reduce signaling risk.  
- Pre-trade liquidity analysis to avoid excessive price impact.


**Regulatory Considerations**
- Rule 144A & Holding Restrictions: Some convertible bonds are issued under Rule 144A, restricting their availability to qualified institutional buyers (QIBs). This may limit the ability to trade certain issues.  
- Short Selling Regulations: Different jurisdictions have varying locate and borrow requirements for short selling the underlying equity. Restrictions on shorting can impact execution and risk management.  
- Position Reporting Requirements: Large convertible arbitrage positions may trigger reporting obligations to regulatory bodies (e.g., SEC 13F filings, EU Short Selling Regulation).  
- Insider Trading & Material Nonpublic Information (MNPI): Convertible bond trading often involves interactions with dealers and corporate disclosures. Firms must ensure compliance with MNPI regulations to avoid legal and reputational risks.  


**Competition**  
Convertible arbitrage has historically been a profitable strategy, attracting institutional players such as hedge funds, proprietary trading firms, and market makers. This creates competition in:  
- Identifying Mispricings: As more firms deploy systematic strategies, the edge from traditional convertible arbitrage may decay over time.  
- Execution Speed: Faster execution and superior market access can provide an advantage in capturing short-lived arbitrage opportunities.  
- Borrow Availability & Costs: Prime brokers may prioritize large hedge funds for stock borrow availability, increasing costs for smaller players.  


To stay competitive, the strategy may need to integrate alternative data sources, machine learning techniques, and superior execution models to maintain an edge in an increasingly efficient market.

```python
# Calculate daily returns
daily_returns = equity_prices.pct_change()

# Compute realized volatility for each ticker, with a rolling window based on TTM
def compute_realized_vol(row):
    ticker = row.name  # Get the ticker name
    ttm_series = ttm_df[ticker]  # Get the TTM values for the given ticker
    rolling_vols = pd.Series(index=ttm_series.index, dtype=float)  # Placeholder for results
    
    for date in ttm_series.index:
        window_size = int(ttm_series.loc[date] * 252)  # Dynamic window size
        if window_size > 1 and date in daily_returns.index:
            idx = daily_returns.index.get_loc(date)  # Locate index of the date
            if idx >= window_size:
                rolling_vols.loc[date] = daily_returns[ticker].iloc[idx - window_size + 1: idx + 1].std() * np.sqrt(252)
    
    return rolling_vols

# Apply function across tickers (columns)
realized_vols = ttm_df.apply(compute_realized_vol, axis=0)
```

### Bond Risk Calculator

The YTM pricer fails in certain scenarios, so in cases where yields are unavailable, we use the average yield of the zero-coupon bond. We consider this approach acceptable since it results in a lower yield estimate, aligning with our conservative modeling approach.

```python
RATE = zcb_rates[5.0].mean()
```

```python
duration_df = pd.DataFrame(index=convert_prices.index, columns=convert_prices.columns)

def bond_yield(price, coupon_rate, years_to_maturity, face_value = 1000, frequency=2):
    """
    Calculate the yield to maturity (YTM) of a bond given its price.

    Parameters:
    - price (float): The current market price of the bond.
    - face_value (float): The bond's face value (typically $100).
    - coupon_rate (float): The annual coupon rate as a decimal (e.g., 0.05 for 5%).
    - years_to_maturity (float): The number of years remaining until maturity.
    - frequency (int, optional): The number of coupon payments per year (default is 2 for semi-annual).

    Returns:
    - float: The yield to maturity (YTM) as a decimal.
    """

    periods = int(years_to_maturity * frequency)  # Total number of coupon payments
    coupon_payment = (coupon_rate * face_value) / frequency  # Coupon per period

    # Function to find the root (YTM)
    def price_function(y):
        discounted_coupons = sum(coupon_payment / (1 + y / frequency) ** (t + 1) for t in range(periods))
        discounted_face_value = face_value / (1 + y / frequency) ** periods
        return discounted_coupons + discounted_face_value - price

    # Initial guess for YTM
    initial_guess = 0.05  # 5% as a starting point

    # Solve for YTM using Newton's method
    try:
        ytm = newton(price_function, initial_guess)
    except RuntimeError:
        
        ytm = RATE

    return ytm

# calculate delta risk associated with each bond at each date
for ticker in convert_prices.columns:
    for date in convert_prices.index:
        # Get the yield to maturity for the current ticker and date
        price = bond_tv.loc[date, ticker]
        cpn = bond_des.loc['Coupon Rate', ticker]
        ttm = ttm_df.loc[date, ticker]
        ytm = bond_yield(price, cpn, ttm)

        duration = helpers.calc_bond_duration(cpn, ytm, ttm, frequency=2)

        # Assign the computed call option price to the DataFrame
        duration_df.loc[date, ticker] = duration
```

## Downside Beta

```python
def downside_beta(asset_returns, market_returns):
    """
    Calculate the downside beta of an asset relative to the market.
    
    Parameters:
    asset_returns (pd.Series): The returns of the asset.
    market_returns (pd.Series): The returns of the market.

    Returns:
    float: Downside beta of the asset.
    """
    # Ensure the data is aligned
    asset_returns, market_returns = asset_returns.align(market_returns, join='inner')

    # Select only negative market returns
    mask = market_returns < 0
    market_downside = market_returns[mask]
    asset_downside = asset_returns[mask]

    # Check if there are enough negative returns to compute downside beta
    if len(market_downside) == 0:
        return np.nan  # Return NaN if no downside movements

    # Compute downside beta
    covariance = np.cov(asset_downside, market_downside)[0, 1]
    variance = np.var(market_downside)

    if variance == 0:
        return np.nan  # Avoid division by zero

    downside_beta_value = covariance / variance
    return downside_beta_value


downside_betas = pd.DataFrame(index=xrets.iloc[:,:3].columns, columns=xrets.iloc[:,3:].columns)
for col in xrets.iloc[:,:3].columns:
    for index in xrets.iloc[:,3:].columns:
        db = downside_beta(xrets[col], xrets[index])
        downside_betas.loc[col, index] = db
    

display(downside_betas.style.format({col: "{:.4f}" for col in downside_betas.select_dtypes(include=['float']).columns}))
```

### Augmented Dickey-Fuller (ADF) Test 

#### 1. Overview 
The ADF test checks whether a time series is **stationary** (i.e., its statistical properties do not change over time).  
- **Null Hypothesis ($H_0$)**: The series has a unit root (non-stationary).  
- **Alternative Hypothesis ($H_1$)**: The series is stationary.  
- **Decision Criteria**:  
  - If the **p-value < 0.05**, reject $H_0$ → The series is stationary.  
  - If the **ADF statistic is very negative**, it suggests strong stationarity.  

#### 2. Results Interpretation  
**Realized PnL**  
- **ADF Statistic = -15.25, p-value = 5.06e-28**  
- **Conclusion**: The p-value is extremely low, strongly rejecting the null hypothesis.  
- **Implication**: Realized PnL is **stationary**, meaning its statistical properties (mean, variance) do not change over time.  

**Unrealized PnL**  
- **ADF Statistic = -22.10, p-value = 0.0**  
- **Conclusion**: The most stationary of the three, with an even lower ADF statistic.  
- **Implication**: Unrealized PnL is **highly stationary**, indicating no persistent trends or drifts.  

**Available Capital**

- **Conclusion**: Still below the 0.05 threshold, meaning it is stationary, but **less so compared to PnL metrics**.  
- **Implication**: Available capital has some fluctuations but remains mean-reverting over time.  

#### 3. Key Takeaways

# Overview

We tested a convertible bond long/short strategy with varying delta re-hedge thresholds  
($\delta_{\text{threshold}} = 0.0005, 0.0015, 0.003, 0.01, 0.03$). The strategy involves:

- Going **long** underpriced convertible bonds (based on a model vs. market price comparison).  
- **Shorting** the underlying equity to hedge delta exposure.  
- **Shorting Treasuries** to hedge duration risk.  

Below is a concise review of the observed realized P&L behavior under each threshold.

$$
\text{Realized P\&L} = \sum_{t} \Bigl(\text{Bond Trade P\&L} + \text{Equity Hedge P\&L} + \text{Treasury Hedge P\&L} - \text{Transaction Costs} \Bigr).
$$


---

## Key Observations  

### 1. Sensitivity to Delta Threshold  
- A *smaller* delta threshold (e.g., 0.0005) triggers more frequent rebalancing.  
- More rebalancing can reduce large swings in equity exposure but increases cumulative transaction costs.  
- Conversely, a *larger* delta threshold (e.g., 0.03) reduces the number of hedge adjustments, potentially saving on costs but exposing the portfolio to more unhedged market moves.  

### 2. Distribution of P&L Spikes  
- For lower thresholds (0.0005, 0.0015), we see frequent, smaller P&L realizations.  
- A notable large negative outlier around mid-2023 suggests a sudden market move that temporarily overwhelmed the hedge adjustments before execution.  
- For higher thresholds (0.01, 0.03), the strategy holds positions longer between re-hedges, resulting in fewer but more pronounced P&L spikes (both positive and negative).  

### 3. Transaction Cost Impact  
- The transaction cost rate ($\text{trans\_cost\_rate} = 0.005$ for bonds, lower for equities/treasuries) significantly influences net P&L.  
- With frequent rebalancing (low delta threshold), these costs accumulate quickly and can erode gains from smaller, more precise hedges.  

### 4. Overall Profitability vs. Risk  
- Across all thresholds, the strategy occasionally captures significant positive gains when convertible bonds revert to fair value.  
- The trade-off between hedging precision and transaction costs becomes evident:  

$$
\text{Expected P\&L} \approx \text{(Bond mispricing capture)} - \text{(Hedging inefficiencies)} - \text{(Transaction costs)}.
$$

- Smaller thresholds help manage equity risk more tightly but can reduce net profitability once costs are factored in.  
- Larger thresholds preserve more capital from fewer transactions but can expose the strategy to larger drawdowns during volatile periods.  

---

## Conclusion  

These plots underscore the delicate balance between *hedge accuracy* and *transaction cost drag*. In practice, choosing an optimal delta threshold depends on:

- **Market Volatility**: Higher volatility often justifies tighter hedging.  
- **Transaction Cost Structure**: Lower costs favor more frequent rehedging.  
- **Capital Constraints**: A comfortable capital buffer is required for frequent adjustments.  

In summary, the strategy can be profitable under a range of delta thresholds. A smaller threshold reduces unhedged exposure but can accumulate substantial costs; a larger threshold lowers trading costs but increases the risk of large swings in P&L. An intermediate threshold (e.g., 0.0015 to 0.003) may offer a more balanced trade-off between hedge effectiveness and transaction costs, though the exact choice should be calibrated to the investor’s risk tolerance and market outlook.

## Key Observations

### 1. Trade Frequency vs. Cheapness Threshold
- A *lower* threshold (e.g., 0.0001) requires the bond to be extremely cheap relative to its model price before triggering a buy. This results in **few trades**, focusing only on the largest mispricings.
- A *higher* threshold (e.g., 0.1) widens the filter, allowing **more trades** to be initiated on smaller mispricings, which can dilute overall returns if transaction costs are significant.

### 2. Realized P&L Distribution
- With **strict** cheapness filters (e.g., 0.0001, 0.02), we observe relatively sparse but occasionally large P&L spikes, suggesting the strategy captures a few, potentially high-margin opportunities.
- **Looser** thresholds (e.g., 0.08, 0.1) yield more frequent trades with moderate P&L outcomes, but we also see the possibility of more negative outliers when market dislocations or hedging costs outweigh smaller mispricing gains.

### 3. Transaction Costs and Capital Utilization
- Each trade incurs transaction costs for convertible bond purchases, equity shorting, and Treasury hedges.
- Under a *high* threshold, these cumulative costs can erode net returns if many trades capture only marginal cheapness differentials.
- Under a *low* threshold, fewer trades mean fewer total transaction costs, but the strategy may sit idle for extended periods.

### 4. Risk-Return Trade-Off
- The $\text{max\_cheapness}$ parameter fundamentally shifts the strategy’s **risk appetite**.
  - A lower threshold targets bigger potential mispricings (thus potentially higher per-trade returns) but fewer opportunities.
  - A higher threshold may provide more stable, incremental gains, albeit with increased market exposure and transaction cost drag.

---

These results illustrate how adjusting the cheapness threshold reshapes the balance between capturing large mispricings and managing transaction costs and exposure.

- **Lower Threshold (e.g., 0.0001, 0.02)**: 
  - Fewer but potentially more lucrative trades.
  - Suitable for investors comfortable waiting for significant market dislocations.
- **Higher Threshold (e.g., 0.08, 0.1)**:
  - More trades on smaller mispricings.
  - Can smooth returns but may be vulnerable to higher cost drag and occasional large drawdowns.



In practice, choosing an optimal cheapness threshold depends on the investor’s **risk tolerance, transaction cost structure, and capital constraints**.
